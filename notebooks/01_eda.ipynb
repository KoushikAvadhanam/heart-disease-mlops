# Heart Disease UCI - Exploratory Data Analysis
Run `python src/download_data.py` first to create `data/heart.csv`.

```python
from pathlib import Path

import matplotlib
import pandas as pd
import seaborn as sns

import matplotlib.pyplot as plt

OUT = Path("screenshots/eda")
OUT.mkdir(parents=True, exist_ok=True)
sns.set_theme(style="whitegrid")
df = pd.read_csv("data/heart.csv")
print(df.shape)
print(df.dtypes)
```

## Missing values and summary statistics

```python
print(df.isna().sum()[df.isna().sum() > 0])
print(df.describe().T)
```

## Class balance

```python
fig, ax = plt.subplots(figsize=(5, 4))
sns.countplot(x="target", data=df, ax=ax)
ax.set_title("Class balance (0 = no disease, 1 = disease)")
for c in ax.containers:
    ax.bar_label(c)
fig.savefig(OUT / "class_balance.png", dpi=120, bbox_inches="tight")
print(df["target"].value_counts(normalize=True).round(3))
```

## Histograms of numeric features, split by target

```python
numeric = ["age", "trestbps", "chol", "thalach", "oldpeak"]
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), numeric):
    sns.histplot(data=df, x=col, hue="target", kde=True, ax=ax, bins=25)
    ax.set_title(f"Distribution of {col}")
axes.ravel()[-1].axis("off")
fig.tight_layout()
fig.savefig(OUT / "histograms.png", dpi=120)
```

## Categorical features vs target

```python
categorical = ["sex", "cp", "fbs", "restecg", "exang", "slope", "ca", "thal"]
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, col in zip(axes.ravel(), categorical):
    sns.countplot(data=df, x=col, hue="target", ax=ax)
    ax.set_title(col)
fig.tight_layout()
fig.savefig(OUT / "categorical_vs_target.png", dpi=120)
```

## Correlation heatmap

```python
fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax)
ax.set_title("Feature correlation heatmap")
fig.savefig(OUT / "correlation_heatmap.png", dpi=120, bbox_inches="tight")
print(df.corr()["target"].drop("target").sort_values(ascending=False).round(3))
```
